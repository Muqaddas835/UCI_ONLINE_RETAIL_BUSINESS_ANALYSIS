```python
df.dropna(subset = ['Description'],inplace=True)
```

```python
df["Customer ID"] = (
    df["Customer ID"]
    .astype("Int64")
    .astype("string")
    .fillna("Missing")
)
```

```python
(df["Quantity"]==0).sum()
```

```python
# Create invoice cancellation condition
cancel = df["Invoice"].astype("string").str.startswith("C", na=False)

# Category 1: Original Sales
sales_data = df[
    (df["Quantity"] > 0) &
    (df["Price"] > 0) &
    (~cancel)
].copy()

# Category 2: Returns / Credit Notes
returns_data = df[
    (df["Quantity"] < 0) &
    (cancel)
].copy()

# Category 3: Adjustments
adjustments_data = df[
    (
        (df["Quantity"] < 0) & (~cancel) & (df["Price"] == 0)
    ) |
    (df["Price"] <= 0)
].copy()
```

```python
print("Sales:", sales_data.shape)
print("Returns:", returns_data.shape)
print("Adjustments:", adjustments_data.shape)
```

```python
# Check overlapping records
sales_index = set(sales_data.index)
returns_index = set(returns_data.index)
adjustments_index = set(adjustments_data.index)

print("Sales & Returns overlap:", len(sales_index & returns_index))
print("Sales & Adjustments overlap:", len(sales_index & adjustments_index))
print("Returns & Adjustments overlap:", len(returns_index & adjustments_index))

# Total categorized records
total_categorized = (
    len(sales_data) +
    len(returns_data) +
    len(adjustments_data)
)

print("Total categorized:", total_categorized)
print("Original dataframe:", len(df))
print("Uncategorized records:", len(df) - total_categorized)
```

```python
# Add category labels
sales_data["Category"] = "Sales"
returns_data["Category"] = "Returns"
adjustments_data["Category"] = "Adjustments"


# Combine all categories
final_data = pd.concat(
    [sales_data, returns_data, adjustments_data, ],
    ignore_index=True
)

print(final_data.shape)
print(final_data["Category"].value_counts())
```

```python
final_data.to_csv(
    "online_retail_cleaned.csv",
    index=False
)
```
